### Q. Implement an Semantic Network of the given relationships

```python
import networkx as nx
G = nx.DiGraph()

relationships = [
    ("Asmit", "Student", "is a"),
    ("Student", "AI", "studies"),
    ("AI", "Data Mining", "includes"),
    ("AI", "Machine Learning", "includes"),
    ("Data Mining", "Python", "uses"),
    ("Machine Learning", "Python", "uses")
]

pos = {
    "Asmit": (0, 3),
    "Student": (1, 3),
    "AI": (2, 2),
    "Data Mining": (1, 0),
    "Machine Learning": (3, 0),
    "Python": (2, -2)
}

for source, target, relation in relationships:
    G.add_edge(source, target, relation=relation)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="lightblue",
    node_size=2500,
)

edge_labels = nx.get_edge_attributes(G, "relation")
nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_size=9
)

plt.title("Semantic Network")
plt.show()
```
