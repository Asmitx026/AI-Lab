### Q. Solve the 8 Puzzle Problem using A* algorithm

```python
import heapq

def heuristic(state: list[list[int]]):
    count = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != 0 and state[i][j] != goal[i][j]:
                count += 1
    return count

def find_blank(state: list[list[int]]):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return (i, j)

def neighbors(state: list[list[int]]):
    (i, j) = find_blank(state)
    moves = ((1,0),(-1,0),(0,1),(0,-1))
    result = []

    for dx,dy in moves:
        nx, ny = i + dx, j + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = [list(row) for row in state]
            new_state[i][j], new_state[nx][ny] = new_state[nx][ny], new_state[i][j]
            result.append(tuple(tuple(row) for row in new_state))
    return result

def astar(state: list[list[int]]):
    pq = []
    heapq.heappush(pq, (heuristic(state), 0, state, [state]))
    visited = set()

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if state == goal:
            return path

        if state in visited:
            continue

        visited.add(state)
        for nxt in neighbors(state):
            new_g = g + 1
            new_f = new_g + heuristic(nxt)
            heapq.heappush(pq, (new_f, new_g, nxt, path + [nxt]))

    return None
```

```python
goal = ((1, 2, 3),
        (4, 5, 6),
        (7, 8, 0)
)

start = ((1, 2, 3),
         (4, 5, 0),
         (7, 6, 8)
)

solution = astar(start)

# Test the neighbors function with the start state
print(f"Start state: {start}")
print(f"Goal state: {goal}")

if solution is not None:
    for step, state in enumerate(solution):
        print(f"Step: {step}")
        for row in state:
            print(row)
        print()
else:
    print("\nNo solution found")
```
